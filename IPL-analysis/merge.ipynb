```python
import pandas as pd
```

```python
match = pd.read_csv('matches - matches.csv')
delivery = pd.read_csv('deliveries.csv.zip')
```

# higest runs in season by ? player

```python
#merge
new_data=delivery.merge(match,left_on='match_id',right_on='id')
```

```python
new_data.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index().drop_duplicates('season').sort_values('season')[['season','batsman']]
```
