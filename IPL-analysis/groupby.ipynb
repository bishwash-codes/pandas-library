```python
# function for the above task
```

```python
def top_runs_against_teams(batsman_name):
    if batsman_name in df['batsman'].unique():
        mask=df['batsman'] == batsman_name
        a=df[mask].groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False)
        return a
    else:
        return 'please input the name of the player properly'
```

```python
top_runs_against_teams('CH gal')
```
