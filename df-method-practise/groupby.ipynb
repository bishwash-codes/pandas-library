# groupby methods and attributes

```python
# find total number of groups
len(imdb.groupby('Genre'))
# or
imdb['Genre'].nunique()
```

```python
# number of items in each group
imdb.groupby('Star1').size().sort_values(ascending=False)
# or
imdb['Star1'].value_counts()
```

```python
# to get all the items of a particular group 
imdb.groupby('Genre').get_group('Action')
```

```python
imdb.groupby('Genre').describe().sort_values(('IMDB_Rating','mean'),ascending=False).head()
```

```python
# agg methods ( usingg dictionary )
imdb.groupby('Genre').agg({
    'Runtime':'mean',
    'IMDB_Rating':['mean','max','min'],
    'No_of_Votes':'sum',
    'Gross':['sum','mean'],
    'Metascore':'mean'})
```

```python
imdb.groupby('Genre').agg(['min','max'])
```

```python
# apply function

# find number of movies starting with X for each group
def movie(groups,x):
    return groups['Series_Title'].str.startswith(x).sum()
imdb.groupby('Genre').apply(movie,'Z')
```

```python
# Find ranking of each movie in the group acc to no. of votes
def movie_rank(groups):
    groups['Votes_Rank'] = groups['No_of_Votes'].rank(ascending = False)
    return groups

imdb.groupby('Genre').apply(movie_rank)
```
