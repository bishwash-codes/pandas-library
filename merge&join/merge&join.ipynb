```python
import numpy as np
import pandas as pd
```

```python
students = pd.read_csv('students.csv') 
courses = pd.read_csv('courses.csv')
november= pd.read_csv('reg-month1.csv')
december= pd.read_csv('reg-month2.csv')
```

```python
#horizontal stacking 
combined_reg = pd.concat([november,december],ignore_index = True)
```

```python
#vertical combining
comb_reg = pd.concat([november,december],axis = 1)
```

```python
# MERGE
```

```python
# inner join
a= students.merge(combined_reg,how='inner',on='student_id')
a.merge(courses,how='inner',on='course_id')
```

```python
# left join
courses.merge(combined_reg,how='left',on='course_id')
```

```python
# right join
combined_reg.merge(courses,how='right',on='course_id')
```

```python
# outer join
students.merge(combined_reg,how='outer',on='student_id')
```

```python
# total revenue in nov and dec through courses
total_rev = combined_reg.merge(courses,how='inner',on='course_id')['price'].sum()
total_rev
```

```python
#month wise revenue
# only input either november or december for now'
def revenue(month):
    return month.merge(courses,on='course_id')['price'].sum()

revenue(november)
```

```python
# find the students who enrolld in both the months
november.merge(december,on='student_id').drop_duplicates('student_id').merge(students,on='student_id').sort_values('student_id')
```

```python
#find courses that got no enrollment
b=combined_reg.merge(courses,how='right',on='course_id')
b[b['student_id'].isnull()]['course_name']
```

```python
#students who are not enroolled in any courses
info = combined_reg.merge(students,how='right',on='student_id')
info[info['course_id'].isnull()]['name']
```

```python
# print student name with their partners but only the enrolled one
# self join 
students.merge(students,how='inner',left_on='partner',right_on='student_id')[['name_x','name_y']].head()
```

```python
# top 5 students with most number of enrollments in diff courses
combined_reg .merge(students,on='student_id').groupby(['student_id','name'])['name'].count().sort_values(ascending=False).head()
```

```python
# find top 5 students with highest expenditure on course
combined_reg .merge(students,on='student_id').merge(courses,on='course_id').groupby(['student_id','name'])['price'].sum().sort_values(ascending= False).head()
```

```python
deliveries =pd.read_csv('deliveries.csv')
matches = pd.read_csv('matches.csv')
```

```python
# top 3 stadiums with highest sixes per match ratio

trial_df = deliveries.merge(matches,left_on='match_id',right_on='id')
six_df = trial_df[trial_df['batsman_runs']== 6]
number_of_six = six_df.groupby('venue')['venue'].count()
number_of_matches = matches['venue'].value_counts()

(number_of_six/number_of_matches).sort_values(ascending=False).head(3)
```

```python
# find orange cap holder of every season
trial_df.groupby(['season','batsman'])['batsman_runs'].sum().reset_index().sort_values('batsman_runs',ascending=False).drop_duplicates('season',keep='first').sort_values('season')
```
